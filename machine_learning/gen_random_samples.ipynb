```python
import os
from glob import glob
from sklearn.model_selection import train_test_split
import numpy as np

np.random.seed(0)
```

```python
root = 'data'
struct_names = os.path.basename(root)
all_strucs = glob(os.path.join(root, '*.xyz')) 
```

```python
all_strucs
```

```python
structures = []
categories = ['ammonia', 'methanol', 'ethylene',
              'ethane', 'methane', 'water', 'formaldehyde']
for category in categories:
    prefix = f"data\\{category}"
    filtered = [s[5:] for s in all_strucs if s.startswith(prefix)]
    sampled = list(np.random.choice(filtered, size=20, replace=False))
    structures.extend(sampled)

print(f"Files: {structures}\n")
print(f"Number of molecules: {len(structures)}")
```

```python
train, test = train_test_split(structures, train_size=0.8, test_size=0.2, random_state=40)

print(f"Train: {train}\n")
print(f"Test: {test}\n")

print(f"# of train molecules: {len(train)}")
print(f"# of test molecules: {len(test)}")
```

```python
for name in train:
    with open('train_names.txt', "a") as f:
        f.write(f"{name}\n")
```

```python
for name in test:
    with open('test_names.txt', "a") as f:
        f.write(f"{name}\n")
```
