```python
from chemml.chem import Molecule
import numpy as np
```

```python
import random

datfile='/mnt/c/Users/Maxim/Documents/Toronto/Research/code/DDLUCJ/gdb11_size04.smi'

random.seed(0)
with open(datfile, 'r') as f:
    lines = f.readlines()
    sample_indices = random.sample(range(len(lines)), 5)   # pick 5 random indices
    sample_lines = [lines[i] for i in sample_indices] 

print(sample_indices)
print(sample_lines)
for idx, i in enumerate(sample_lines):
    molex = Molecule(i.strip(), input_type='smiles')
    molex.hydrogens('add')
    molex.to_xyz(optimizer='MMFF', mmffVariant='MMFF94s', maxIters=300)
    dummyarray = np.append(molex.xyz.atomic_symbols, molex.xyz.geometry, axis=1)

    sample_idx = sample_indices[idx]
    with open(f'GDB11/GDB04_{sample_idx}.xyz', 'w') as xyzfilename:
        xyzfilename.write(f'{len(dummyarray)} \n\n')
        for a, x, y, z in dummyarray:
            x1, y1, z1 = ['{:.6f}'.format(e) for e in (float(x), float(y), float(z))]
            xyzfilename.write('{} {:>13} {:>13} {:>13} \n'.format(a, x1, y1, z1))
```
