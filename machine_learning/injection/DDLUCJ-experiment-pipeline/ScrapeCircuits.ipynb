```python
transpilationstats = []
for jobs in glob("jobids/*txt"):
    file = np.genfromtxt(jobs,dtype=str)
    name, basis, injection, layers, jobid = file

    job = service.job(jobid)
    job_result = job.result()
    isa_circuit = job.inputs['pubs'][0][0]

    depth = isa_circuit.depth()
    
    two_qubit_depth = isa_circuit.depth(lambda instr: len(instr.qubits)==2)
    two_qubit_count = isa_circuit.num_nonlocal_gates()    
    
    transpilationstats.append((name, basis, injection, layers, jobid, depth, two_qubit_depth, two_qubit_count))
```

```python
statsdf = pd.DataFrame(transpilationstats,columns=['Name','Basis Set','Injection','L','JobID','Circuit Depth','Two-Qubit Depth','Two-Qubit Count']).drop(columns="JobID")
# statsdf    .to_excel("circuitdepth.xlsx")
```
