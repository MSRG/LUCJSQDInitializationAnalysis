```python
import numpy as np
from glob import glob

import os, shutil
```

```python
for dirs in glob("./*"):
    if os.path.isdir(dirs):
        basisset = dirs.replace('./','')

        for i in glob(f"./{basisset}/amplitudes/*_{basisset}.npz"):
            file_size_bytes = os.path.getsize(i)
            file_size_mb = file_size_bytes / (1024 * 1024)
            # print(f"The size of '{i}' is {file_size_bytes} bytes.")
            print(f"The size of '{i}' is {file_size_mb:.2f} MB.")
            
            with open(i,'rb') as f:
                loaded = np.load(f)
                for k,v in dict(loaded).items():
                    small_name = i.replace('.npz','')+f'_{k}.npz'
        
                    with open(small_name,'wb') as g:
                        np.savez_compressed(g,k=v)        
```

```python
os.mkdir('moved_amps')
for dirs in glob("./*"):
    if os.path.isdir(dirs):
        basisset = dirs.replace('./','')

        for i in glob(f"./{basisset}/amplitudes/*_{basisset}.npz"):
            print(i)
            shutil.move(i,'moved_amps')
```
